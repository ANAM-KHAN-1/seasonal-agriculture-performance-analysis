```python
# Create a copy of the original Yield column for audit purposes

df["Yield_Tonnes_Ha_Original"] = df["Yield_Tonnes_Ha"]

# Calculate yield from production and farm area
calculated_yield = (
    df["Production_Tonnes"] /
    df["Farm_Area_Hectares"]
)

# Fill only missing Yield values using the calculated relationship
missing_yield_mask = df["Yield_Tonnes_Ha"].isna()

df.loc[missing_yield_mask, "Yield_Tonnes_Ha"] = (
    calculated_yield[missing_yield_mask].round(2)
)

print("Missing Yield values before cleaning:", missing_yield_mask.sum())
print(
    "Missing Yield values after cleaning:",
    df["Yield_Tonnes_Ha"].isna().sum()
)
```

```python
# Check the number of observations available for each
# State + Season + Crop combination

group_size = (
    df.groupby(["State", "Season", "Crop"])
      .size()
      .reset_index(name="Record_Count")
      .sort_values("Record_Count")
)

group_size.head(20)
```

```python
# Handle missing Rainfall and Soil Moisture values
# using hierarchical median imputation

imputation_columns = [
    "Rainfall_mm",
    "Soil_Moisture_pct"
]

# Keep original values for auditability
for column in imputation_columns:
    df[f"{column}_Original"] = df[column]

# --------------------------------------------------
# Level 1: State + Season + Crop median
# --------------------------------------------------

for column in imputation_columns:
    group_median = (
        df.groupby(["State", "Season", "Crop"])[column]
        .transform("median")
    )
    
    df[column] = df[column].fillna(group_median)

# --------------------------------------------------
# Level 2: State + Season median
# --------------------------------------------------

for column in imputation_columns:
    group_median = (
        df.groupby(["State", "Season"])[column]
        .transform("median")
    )
    
    df[column] = df[column].fillna(group_median)

# --------------------------------------------------
# Level 3: Season median
# --------------------------------------------------

for column in imputation_columns:
    season_median = (
        df.groupby("Season")[column]
        .transform("median")
    )
    
    df[column] = df[column].fillna(season_median)

# --------------------------------------------------
# Level 4: Overall median
# --------------------------------------------------

for column in imputation_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing values after imputation:")
print(df[imputation_columns].isna().sum())
```

```python
# Verify missing values after data cleaning

missing_after_cleaning = df.isnull().sum()

missing_after_cleaning = (
    missing_after_cleaning[missing_after_cleaning > 0]
    .sort_values(ascending=False)
)

if len(missing_after_cleaning) == 0:
    print("✓ No missing values remain in the dataset.")
else:
    print("Remaining missing values:")
    print(missing_after_cleaning)
```

```python
# Detect potential outliers in numerical variables using the IQR method

numerical_columns = df.select_dtypes(include=np.number).columns

outlier_summary = []

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": outlier_count,
        "Outlier_Percentage": round(
            (outlier_count / len(df)) * 100, 2
        )
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    "Outlier_Count",
    ascending=False
)
```

```python
# Define analytical numerical columns
# Exclude audit/original columns created during cleaning

analytical_numeric_columns = [
    col for col in df.select_dtypes(include=np.number).columns
    if not col.endswith("_Original")
]

print("Number of analytical numerical variables:",
      len(analytical_numeric_columns))

print("\nAnalytical numerical variables:")
print(analytical_numeric_columns)
```

```python
# Create IQR-based outlier flags for analytical variables

outlier_flags = pd.DataFrame(index=df.index)

for column in analytical_numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_flags[column] = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    )

# Count how many variables identify each record as an outlier
df["Outlier_Variable_Count"] = outlier_flags.sum(axis=1)

print("Records with at least one potential outlier:",
      (df["Outlier_Variable_Count"] > 0).sum())

print("Records with 3 or more potential outlier variables:",
      (df["Outlier_Variable_Count"] >= 3).sum())
```

```python
# Inspect records with multiple potential outlier variables

multi_outlier_records = df[
    df["Outlier_Variable_Count"] >= 3
].copy()

# Show the most extreme records first
multi_outlier_records[
    [
        "Farm_ID",
        "State",
        "District",
        "Crop",
        "Season",
        "Farm_Area_Hectares",
        "Yield_Tonnes_Ha",
        "Production_Tonnes",
        "Revenue_INR",
        "Profit_INR",
        "Water_Used_m3",
        "Water_Efficiency_t_per_1000m3",
        "Outlier_Variable_Count"
    ]
].sort_values(
    "Outlier_Variable_Count",
    ascending=False
).head(20)
```

```python
# Validate important business relationships

# 1. Production = Farm Area × Yield
df["Calculated_Production"] = (
    df["Farm_Area_Hectares"] * df["Yield_Tonnes_Ha"]
)

df["Production_Difference"] = (
    df["Production_Tonnes"] -
    df["Calculated_Production"]
).abs()


# 2. Revenue = Production × Market Price
df["Calculated_Revenue"] = (
    df["Production_Tonnes"] *
    df["Market_Price_INR_Tonne"]
)

df["Revenue_Difference"] = (
    df["Revenue_INR"] -
    df["Calculated_Revenue"]
).abs()


# 3. Profit = Revenue − Total Cost
df["Calculated_Profit"] = (
    df["Revenue_INR"] -
    df["Total_Cost_INR"]
)

df["Profit_Difference"] = (
    df["Profit_INR"] -
    df["Calculated_Profit"]
).abs()


print("Production mismatches (> 0.1):",
      (df["Production_Difference"] > 0.1).sum())

print("Revenue mismatches (> 1):",
      (df["Revenue_Difference"] > 1).sum())

print("Profit mismatches (> 1):",
      (df["Profit_Difference"] > 1).sum())
```

```python
# Create a clean analytical dataset
# Remove temporary audit and validation columns

temporary_columns = [
    "Rainfall_mm_Original",
    "Soil_Moisture_pct_Original",
    "Yield_Tonnes_Ha_Original",
    "Outlier_Variable_Count",
    "Calculated_Production",
    "Production_Difference",
    "Calculated_Revenue",
    "Revenue_Difference",
    "Calculated_Profit",
    "Profit_Difference"
]

df_clean = df.drop(columns=temporary_columns, errors="ignore").copy()

print("Original dataset shape :", df.shape)
print("Clean dataset shape   :", df_clean.shape)
```

##  RESULT 1 — Seasonal Agricultural Performance

**Objective:** Compare agricultural performance across Kharif, Rabi, and Zaid seasons using average yield and production.

**Key Focus:**
-  Highest and lowest average yield
-  Seasonal production differences
-  Percentage variation in yield
-  Best-performing season

```python

season_result = (
    df_clean
    .groupby("Season")
    .agg(
        Avg_Yield=("Yield_Tonnes_Ha", "mean"),
        Avg_Production=("Production_Tonnes", "mean"),
        Farm_Count=("Farm_ID", "count")
    )
    .sort_values("Avg_Yield", ascending=False)
)

# Chart
fig, ax1 = plt.subplots(figsize=(10, 6))

x = np.arange(len(season_result))

ax1.bar(
    x - 0.2,
    season_result["Avg_Yield"],
    width=0.4,
    label="Average Yield (Tonnes/Ha)"
)

ax2 = ax1.twinx()

ax2.bar(
    x + 0.2,
    season_result["Avg_Production"],
    width=0.4,
    label="Average Production (Tonnes)"
)

ax1.set_xticks(x)
ax1.set_xticklabels(season_result.index)

ax1.set_xlabel("Season")
ax1.set_ylabel("Average Yield (Tonnes/Ha)")
ax2.set_ylabel("Average Production (Tonnes)")

plt.title("RESULT 1 — Seasonal Agricultural Performance")

plt.tight_layout()
plt.show()


# Findings
best_yield_season = season_result["Avg_Yield"].idxmax()
worst_yield_season = season_result["Avg_Yield"].idxmin()

best_yield = season_result.loc[
    best_yield_season, "Avg_Yield"
]

worst_yield = season_result.loc[
    worst_yield_season, "Avg_Yield"
]

yield_difference_pct = (
    (best_yield - worst_yield) /
    worst_yield
) * 100

best_production_season = season_result[
    "Avg_Production"
].idxmax()

print("KEY OBSERVATIONS")
print(
    f"1. Highest average yield: {best_yield_season} "
    f"({best_yield:.2f} Tonnes/Ha)"
)

print(
    f"2. Lowest average yield: {worst_yield_season} "
    f"({worst_yield:.2f} Tonnes/Ha)"
)

print(
    f"3. Yield difference between highest and lowest "
    f"season: {yield_difference_pct:.2f}%"
)

print(
    f"4. Highest average production: "
    f"{best_production_season}"
)
```

##  RESULT 2 — Crop-wise Seasonal Trends

**Objective:** Analyze how crop yield and production vary across different seasons.

**Key Focus:**
-  Highest-yielding crop–season combination
-  Lowest-yielding crop–season combination
-  Seasonal yield trends across crops
-  Overall highest-performing crop

```python
# ============================================================
# RESULT 2 — CROP / PRODUCTION RELATED SEASONAL TRENDS
# ============================================================

crop_season_result = (
    df_clean
    .groupby(["Season", "Crop"])
    .agg(
        Avg_Yield=("Yield_Tonnes_Ha", "mean"),
        Avg_Production=("Production_Tonnes", "mean")
    )
    .reset_index()
)

# Pivot for chart
yield_pivot = crop_season_result.pivot(
    index="Crop",
    columns="Season",
    values="Avg_Yield"
)

# Chart
yield_pivot.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("RESULT 2 — Crop-wise Seasonal Yield Trends")
plt.xlabel("Crop")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=0)
plt.legend(title="Season")
plt.tight_layout()
plt.show()


# Best crop-season combination
best_combination = crop_season_result.loc[
    crop_season_result["Avg_Yield"].idxmax()
]

worst_combination = crop_season_result.loc[
    crop_season_result["Avg_Yield"].idxmin()
]

print("KEY OBSERVATIONS")

print(
    f"1. Highest-yielding crop-season combination: "
    f"{best_combination['Crop']} in "
    f"{best_combination['Season']} "
    f"({best_combination['Avg_Yield']:.2f} Tonnes/Ha)"
)

print(
    f"2. Lowest-yielding crop-season combination: "
    f"{worst_combination['Crop']} in "
    f"{worst_combination['Season']} "
    f"({worst_combination['Avg_Yield']:.2f} Tonnes/Ha)"
)

# Number of crop-season combinations
print(
    f"3. Total crop-season combinations analysed: "
    f"{len(crop_season_result)}"
)

# Overall best crop
overall_crop = (
    df_clean
    .groupby("Crop")["Yield_Tonnes_Ha"]
    .mean()
    .idxmax()
)

print(
    f"4. Highest average-yielding crop overall: "
    f"{overall_crop}"
)
```

##  RESULT 3 — Resource Usage Across Seasons

**Objective:** Compare water, fertilizer, and land usage across different seasons and examine their relationship with production.

**Key Focus:**
-  Highest average water usage
-  Highest fertilizer usage
-  Highest average production
-  Relationship between water usage and production

```python
# ============================================================
# RESULT 3 — RESOURCE USAGE ACROSS SEASONS
# ============================================================

resource_result = (
    df_clean
    .groupby("Season")
    .agg(
        Avg_Water_Used=("Water_Used_m3", "mean"),
        Avg_Fertilizer=("Fertilizer_kg_ha", "mean"),
        Avg_Farm_Area=("Farm_Area_Hectares", "mean"),
        Avg_Production=("Production_Tonnes", "mean")
    )
    .round(2)
)

# Normalize resources for comparison
resource_index = resource_result[
    [
        "Avg_Water_Used",
        "Avg_Fertilizer",
        "Avg_Farm_Area"
    ]
].copy()

resource_index = (
    resource_index /
    resource_index.max()
)

# Chart
resource_index.plot(
    kind="bar",
    figsize=(11, 6)
)

plt.title("RESULT 3 — Relative Resource Usage Across Seasons")
plt.xlabel("Season")
plt.ylabel("Relative Resource Usage")
plt.xticks(rotation=0)
plt.legend(
    title="Resource",
    labels=[
        "Water Used",
        "Fertilizer",
        "Farm Area"
    ]
)

plt.tight_layout()
plt.show()


# Findings
highest_water_season = resource_result[
    "Avg_Water_Used"
].idxmax()

highest_fertilizer_season = resource_result[
    "Avg_Fertilizer"
].idxmax()

highest_production_season = resource_result[
    "Avg_Production"
].idxmax()

water_production_corr = df_clean[
    ["Water_Used_m3", "Production_Tonnes"]
].corr().iloc[0, 1]

print("KEY OBSERVATIONS")

print(
    f"1. Highest average water usage: "
    f"{highest_water_season}"
)

print(
    f"2. Highest average fertilizer usage: "
    f"{highest_fertilizer_season}"
)

print(
    f"3. Highest average production: "
    f"{highest_production_season}"
)

print(
    f"4. Correlation between water usage and production: "
    f"{water_production_corr:.3f}"
)
```

###  RESULT 4 — Environmental Factors

-  Identified the strongest environmental factor affecting yield
-  Calculated Pearson correlation with crop yield
-  Analyzed positive and negative relationships
-  Visualized the strongest factor vs yield

```python
# ============================================================
# RESULT 4 — ENVIRONMENTAL FACTORS VS AGRICULTURAL PERFORMANCE
# ============================================================

environment_columns = [
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_pH",
    "Soil_Moisture_pct"
]

environment_corr = (
    df_clean[
        environment_columns + ["Yield_Tonnes_Ha"]
    ]
    .corr()["Yield_Tonnes_Ha"]
    .drop("Yield_Tonnes_Ha")
    .sort_values(
        key=abs,
        ascending=False
    )
)

print("Environmental factor correlations with yield:\n")
print(environment_corr.round(3))


# Strongest relationship
strongest_factor = environment_corr.index[0]
strongest_corr = environment_corr.iloc[0]


# Scatter plot for strongest factor
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df_clean,
    x=strongest_factor,
    y="Yield_Tonnes_Ha",
    alpha=0.6
)

plt.title(
    f"RESULT 4 — {strongest_factor} vs Crop Yield"
)

plt.xlabel(strongest_factor)
plt.ylabel("Yield (Tonnes/Ha)")

plt.tight_layout()
plt.show()


# Findings
print("\n KEY OBSERVATIONS")

print(
    f"1. Strongest environmental relationship with yield: "
    f"{strongest_factor}"
)

print(
    f"2. Pearson correlation: "
    f"{strongest_corr:.3f}"
)

positive_factors = environment_corr[
    environment_corr > 0
]

negative_factors = environment_corr[
    environment_corr < 0
]

print(
    f"3. Environmental factors with positive association: "
    f"{len(positive_factors)}"
)

print(
    f"4. Environmental factors with negative association: "
    f"{len(negative_factors)}"
)
```

###  RESULT 5 — Economic Performance

-  Identified the most profitable season
-  Compared revenue, cost & profit across seasons
-  Identified the best season by yield
-  Identified the most water-efficient season

```python
# ============================================================
# RESULT 5 — ECONOMIC PERFORMANCE + OVERALL FINDINGS
# ============================================================

economic_result = (
    df_clean
    .groupby("Season")
    .agg(
        Avg_Revenue=("Revenue_INR", "mean"),
        Avg_Cost=("Total_Cost_INR", "mean"),
        Avg_Profit=("Profit_INR", "mean")
    )
    .sort_values(
        "Avg_Profit",
        ascending=False
    )
)

# Chart
economic_result.plot(
    kind="bar",
    figsize=(11, 6)
)

plt.title("RESULT 5 — Economic Performance by Season")
plt.xlabel("Season")
plt.ylabel("Average Value (INR)")
plt.xticks(rotation=0)
plt.legend(
    title="Economic Metric",
    labels=[
        "Revenue",
        "Cost",
        "Profit"
    ]
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Overall findings
# ------------------------------------------------------------

best_profit_season = economic_result[
    "Avg_Profit"
].idxmax()

worst_profit_season = economic_result[
    "Avg_Profit"
].idxmin()

best_profit = economic_result.loc[
    best_profit_season,
    "Avg_Profit"
]

worst_profit = economic_result.loc[
    worst_profit_season,
    "Avg_Profit"
]

profit_difference_pct = (
    (best_profit - worst_profit) /
    abs(worst_profit)
) * 100


best_overall_yield_season = (
    df_clean
    .groupby("Season")["Yield_Tonnes_Ha"]
    .mean()
    .idxmax()
)

best_water_efficiency_season = (
    df_clean
    .groupby("Season")[
        "Water_Efficiency_t_per_1000m3"
    ]
    .mean()
    .idxmax()
)

best_revenue_season = economic_result[
    "Avg_Revenue"
].idxmax()


print("\n" + "=" * 65)
print("              FINAL PROJECT FINDINGS")
print("=" * 65)

print(
    f"\n Best performing season by yield: "
    f"{best_overall_yield_season}"
)

print(
    f" Most profitable season: "
    f"{best_profit_season} "
    f"(Average Profit = ₹{best_profit:,.2f})"
)

print(
    f" Lowest average profit season: "
    f"{worst_profit_season} "
    f"(Average Profit = ₹{worst_profit:,.2f})"
)

print(
    f" Profit difference between best and lowest season: "
    f"{profit_difference_pct:.2f}%"
)

print(
    f" Best water-efficiency season: "
    f"{best_water_efficiency_season}"
)

print(
    f" Highest average revenue season: "
    f"{best_revenue_season}"
)

print(
    f" Strongest environmental factor associated with yield: "
    f"{strongest_factor} "
    f"(r = {strongest_corr:.3f})"
)

print("\n" + "=" * 65)
```
